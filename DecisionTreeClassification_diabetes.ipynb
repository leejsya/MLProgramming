# 1. 라이브러리 및 모듈 import

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import classification_report
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
```

# 2. DataFrame 가져오기

```python
df = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/diabetes.csv")
df
```

# 3. input, target variable 설정

```python
X = df.drop('Outcome', axis=1)
y = df[['Outcome']]
```

# 4. train, test set 분할

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=5)
```

# 5. model 선언 및 training

```python
# DecisionTree model
DTC_model = DecisionTreeClassifier()
DTC_model.fit(X_train, y_train)

# RandomForest model
RFC_model = RandomForestClassifier()
RFC_model.fit(X_train, y_train)

# LogisticRegression model
LR_model = LogisticRegression(max_iter=200)
LR_model.fit(X_train, y_train)

# SVC model
SVC_model = SVC()
SVC_model.fit(X_train, y_train)

# KNN model
KNN_model = KNeighborsClassifier()
KNN_model.fit(X_train, y_train)
```

# 6. model로 test_set 예측

```python
y_pred_DTC = DTC_model.predict(X_test)

y_pred_RFC = RFC_model.predict(X_test)

y_pred_LR = LR_model.predict(X_test)

y_pred_SVC = SVC_model.predict(X_test)
```

# 7. model 평가지표

```python
print("DecisionTree:", classification_report(y_test, y_pred_DTC))
print("RandomForest:", classification_report(y_test, y_pred_RFC))
print("LogisticRegression:", classification_report(y_test, y_pred_LR))
print("SVC:", classification_report(y_test, y_pred_SVC))
```
