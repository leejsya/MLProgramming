### 16. IQR을 이용한 이상값 탐지
IQR = 3사분위수 - 1사분위수

```python
# outliers 계산하는 함수
def OUTLIER(col):
  Q1 = df[col].quantile(0.25)
  Q3 = df[col].quantile(0.75)
  IQR = Q3 - Q1
  lower = Q1 - 1.5 * IQR
  upper = Q3 + 1.5 * IQR

  # lower보다 작거나 upper보다 큰 값을 이상값으로 간주
  outliers = df[
      (df[col] < lower) | (df[col] > upper)
  ]
  return outliers, lower, upper

for col in numeric_cols:
  outliers, lower, upper = OUTLIER(col)
  print(f"{col:20s} : {len(outliers)}개 (Lower={lower:.2f}, Upper={upper:.2f})")
```

```python
# 이상값 직접 확인
for col in numeric_cols:
  outliers, _, _ = OUTLIER(col)
  if len(outliers) > 0:
    print(f"\n[{col} 이상값]")
    display(outliers[[col, target_col]])
```

### 17. 품종별 평균값 비교

```python
# normalization을 위해 min-max scaling 수행
df_normalized = df
df_normalized[numeric_cols] = (
    (df[numeric_cols] - df[numeric_cols].min()) / (df[numeric_cols].max() - df[numeric_cols].min())
)
mean_long = (
    df_normalized.groupby(target_col)[numeric_cols]
      .mean()
      .reset_index()
      .melt(
          id_vars=target_col,
          var_name="Feature",
          value_name="Mean"
      )
)

plt.figure(figsize=(10, 5))

sns.barplot(
    data=mean_long,
    x="Feature",
    y="Mean",
    hue=target_col
)

plt.title("Mean Feature Values by Varieties")
plt.xlabel("Feature")
plt.ylabel("Mean")

plt.xticks(rotation=20)

plt.tight_layout()
plt.show()
```

### 18. EDA 해석

- 13개의 연속형 변수, 하나의 품종 클래스형 변수가 있다.
- 제공된 데이터셋의 총 데이터 개수는 178개이다.
- 데이터 전처리 과정에서 결측값과 중복값은 발견되지 않았다.
- target variable의 class간 데이터 불균형은 일어날 가능성이 적지만, accuracy 이외에 f1-score 지표도 같이 고려하는 것이 적절해 보인다.
- 각 13개의 연속형 변수의 값의 단위와 분포가 다르긱 때문에 normalization을 통해 변수의 unit size를 조정하는 과정이 필요해 보인다.
- Malic.acid 변수의 데이터 분포에서 특정값일 때 frequency가 특징적으로 증가하는 것으로 보아 class를 분류하기 위한 기준으로 설정하기에는 어렵다고 판단된다.
- class별 데이터 분포 그래프를 보았을 때 Ash, Acl, Mg, Proanth 레이블이 세 클래스를 분류하는 특징(영역)을 추출하지 못한다고 판단된다.
- Malic.acid, Mg, Proanth 레이블에서에서 데이터셋 전체에 영향을 줄만한 이상값이 많이 발견되었다. 따라서 해당 레이블에서 이상값은 모두 제거하고, 다른 레이블에 존재하는 이상값은 적절한 대체값을 찾는 것이 적절해보인다.
- Flavanoids, Phenols, OD 세 레이블이 서로 강한 양의 상관관계를 보인다.
- 위 내용을 종합적으로 고려해봤을 때 Alcohol, Phenols, Flavanoids, Nonflavanoids.phenols, Color.int, Hue, OD, Proline 레이블에서 class를 분류하는 특징을 효과적으로 추출할 수 있을 것으로 보인다.
