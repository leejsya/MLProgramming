```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier

from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report
```

```python
df = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/car+evaluation/car.data',
                 names = [
                     'buying', 'maint', 'doors', 'persons', 'lug_boot', 'safety', 'eval'
                 ])
```

```python
df
```

```python
le = LabelEncoder()
for column in df.columns:
    df[column] = le.fit_transform(df[column])
```

```python
X = df.drop('eval', axis=1)
y = df['eval']

print(X)
print("\n")
print(y)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=5)
```

```python
LR_model = LogisticRegression()
RF_model = RandomForestClassifier()
DT_model = DecisionTreeClassifier()
KNN_model = KNeighborsClassifier()
```

```python
LR_model.fit(X_train, y_train)
RF_model.fit(X_train, y_train)
DT_model.fit(X_train, y_train)
KNN_model.fit(X_train, y_train)
```

```python
LR_y_pred = LR_model.predict(X_test)
RF_y_pred = RF_model.predict(X_test)
DT_y_pred = DT_model.predict(X_test)
KNN_y_pred = KNN_model.predict(X_test)

```

```python
print(f" LR:\n{classification_report(y_test, LR_y_pred)}")
print(f" RF:\n{classification_report(y_test, RF_y_pred)}")
print(f" DT:\n{classification_report(y_test, DT_y_pred)}")
print(f" KNN:\n{classification_report(y_test, KNN_y_pred)}")
```
