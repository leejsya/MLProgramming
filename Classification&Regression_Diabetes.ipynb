# Classification - Diabetes Dataset

### 1. 라이브러리 및 모듈 import

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import classification_report
from sklearn.ensemble import RandomForestClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
```

### 2. Dataset 불러오기

```python
df = pd.read_csv("/content/drive/MyDrive/Colab Notebooks/diabetes.csv")
df
```

### 3. input, target variable 설정

```python
# input
X1 = df.drop('Outcome', axis=1)

# target
y1 = df['Outcome']

print(f"X1\n {X1}\n")
print(f"y1\n {y1}")
```

### 4. train, test set 분할

```python
# train : test = 0.8 : 0.2 로 설정.
X1_train, X1_test, y1_train, y1_test = train_test_split(X1, y1, test_size=0.2, random_state=5)
```

### 5. model 선언 및 training

```python
# DecisionTree model
DTC_model = DecisionTreeClassifier()
DTC_model.fit(X1_train, y1_train)

# RandomForest model
RFC_model = RandomForestClassifier()
RFC_model.fit(X1_train, y1_train)

# LogisticRegression model
LogisticR_model = LogisticRegression(max_iter=200)
LogisticR_model.fit(X1_train, y1_train)

# KNN model
KNN_model = KNeighborsClassifier()
KNN_model.fit(X1_train, y1_train)
```

### 6. test set 예측

```python
# DecisionTree model 예측
y1_pred_DTC = DTC_model.predict(X1_test)

# RandomForest model 예측
y1_pred_RFC = RFC_model.predict(X1_test)

# LogisticRegression model 예측
y1_pred_LogisticR = LogisticR_model.predict(X1_test)

# KNN model 예측
y1_pred_KNN = KNN_model.predict(X1_test)
```

### 7. model 평가지표

```python
print(f"DecisionTree: \n{classification_report(y1_test, y1_pred_DTC)}\n")
print(f"RandomForest: \n{classification_report(y1_test, y1_pred_RFC)}\n")
print(f"LogisticRegression: \n{classification_report(y1_test, y1_pred_LogisticR)}\n")
print(f"KNN: \n{classification_report(y1_test, y1_pred_KNN)}")
```

# Regression - Diabetes Dataset

### 1. 라이브러리 및 모듈 import

```python
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
```

### 2. Outcome 레이블을 삭제하고 BMI 레이블을 target variable로 설정.

```python
# input
X2 = df.drop(['Outcome', 'BMI'], axis=1)

# target
y2 = df['BMI']

print(f"X2\n {X2}\n")
print(f"y2\n {y2}")
```

### 3. train, test set 분할

```python
# train : test = 0.8 : 0.2 로 설정.
X2_train, X2_test, y2_train, y2_test = train_test_split(X2, y2, test_size=0.2, random_state=5)
```

### 4. model 선언 및 training

```python
# DecisionTree model
DTR_model = DecisionTreeRegressor()
DTR_model.fit(X2_train, y2_train)

# RandomForest model
RFR_model = RandomForestRegressor()
RFR_model.fit(X2_train, y2_train)

# LinearRegression model
LR_model = LinearRegression()
LR_model.fit(X2_train, y2_train)

```

### 5. test set 예측

```python
# DecisionTree model 예측
y2_pred_DTR = DTR_model.predict(X2_test)

# RandomForest model 예측
y2_pred_RFR = RFR_model.predict(X2_test)

# LinearRegression model 예측
y2_pred_LR = LR_model.predict(X2_test)
```

### 6. model 평가지표

```python
print(f"DecisionTree: \n{mean_squared_error(y2_pred_DTR, y2_test)}\n")
print(f"RandomForest: \n{mean_squared_error(y2_pred_RFR, y2_test)}\n")
print(f"LinearRegression: \n{mean_squared_error(y2_pred_LR, y2_test)}\n")
```
