```python
import pandas as pd

from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/auto+mpg/auto-mpg.data-original',
                 sep='\s+',
                 names=[
                     'mpg', 'cylinders','displacement','horsepower','weight','acceleration','model year','origin','car name'
                 ])
```

```python
df.head()
```

```python
df = df.dropna()
X = df.drop(['origin','car name', 'mpg'], axis=1)
y = df['mpg']

print(X)
print('\n')
print(y)
```

```python
df.isna().sum()
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=5)
```

```python
DT_model = DecisionTreeRegressor()
RF_model = RandomForestRegressor()
LR_model = LinearRegression()
```

```python
DT_model.fit(X_train, y_train)
RF_model.fit(X_train, y_train)
LR_model.fit(X_train, y_train)
```

```python
DT_y_pred = DT_model.predict(X_test)
RF_y_pred = RF_model.predict(X_test)
LR_y_pred = LR_model.predict(X_test)
```

```python
print(f"DT:\n {mean_squared_error(DT_y_pred, y_test)}\n")
print(f"RF:\n {mean_squared_error(RF_y_pred, y_test)}\n")
print(f"LR:\n {mean_squared_error(LR_y_pred, y_test)}\n")
```
